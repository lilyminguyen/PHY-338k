Gain vs. Frequency (gain=100)

```python
f100 = np.array([
    13.48, 19.84, 56.56, 176.8, 289.3, 534.1, 762.1, 1016.0, 2097.0, 7568.0,
    13810.0, 30940.0, 57600.0, 71020.0, 100400.0, 204900.0, 322500.0,
    676800.0, 822300.0, 1015000.0
])

# fake data for vin and vout

vin100 = np.array([
    0.0390, 0.0426, 0.0340, 0.0376, 0.0358,
    0.0351, 0.0362, 0.0381, 0.0388, 0.0405,
    0.0390, 0.0422, 0.0365, 0.0445, 0.0343,
    0.0420, 0.0390, 0.0407, 0.0357, 0.0364
])

vout100 = np.array([
    4.252, 4.863, 3.210, 3.980, 3.980,
    3.926, 3.172, 3.287, 3.492, 4.506,
    3.433, 3.997, 2.159, 1.900, 1.085,
    0.582, 0.383, 0.199, 0.111, 0.116
]) 

gain100 = abs(vout100 / vin100)

plt.figure(figsize=(6,4))
plt.scatter(f100, gain100, color="#97DBF2", s=25, label='Ratio=100')
plt.scatter(f10, gain10, color="#F5BFE6", s=25, label='Ratio=10')
plt.xscale('log')
plt.yscale('log')
plt.xlabel('Frequency (Hz)')
plt.ylabel(r'Gain $\left|\frac{V_{\text{out}}}{V_{\text{in}}}\right|$')
plt.title('Gain vs. Frequency for Inverting Amplifier')
plt.grid(alpha=0.4, linestyle='--')
plt.tight_layout()
plt.legend()
plt.savefig('7.2.a.png')
plt.show()

```
