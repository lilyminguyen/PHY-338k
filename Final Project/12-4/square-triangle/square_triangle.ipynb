# Square Triangle

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.read_csv('square-triangle_R2_2k.CSV', skiprows=15)
df.head()

# both
plt.figure(figsize=(8,4))
plt.plot(df['TIME'], df['CH1'], color='#A2C2E0', linewidth=1.4)
plt.plot(df['TIME'], df['CH2'], color="#FFBFF9", linewidth=1.4)
plt.xlabel('Time (s)', fontsize=11)
plt.ylabel('Voltage (V)', fontsize=11)
plt.grid(True, linestyle=':', alpha=0.6)
plt.tick_params(axis='both', labelsize=10)
plt.tight_layout()
plt.title('square-triangle')
plt.show()

# square only
plt.figure(figsize=(8,4))
plt.plot(df['TIME'], df['CH1'], color='#A2C2E0', linewidth=1.4)
plt.scatter(df['TIME'], df['CH1'], color='#A2C2E0', linewidth=1.4)
# plt.plot(df['TIME'], df['CH2'], color="#FFBFF9", linewidth=1.4)
plt.xlabel('Time (s)', fontsize=11)
plt.ylabel('Voltage (V)', fontsize=11)
plt.grid(True, linestyle=':', alpha=0.6)
plt.tick_params(axis='both', labelsize=10)
plt.tight_layout()
plt.title('square-triangle')
plt.show()

# triangle only
plt.figure(figsize=(8,4))
# plt.plot(df['TIME'], df['CH1'], color='#A2C2E0', linewidth=1.4)
plt.plot(df['TIME'], df['CH2'], color="#FFBFF9", linewidth=1.4)
plt.scatter(df['TIME'], df['CH2'], color="#FFBFF9", linewidth=1.4)
plt.xlabel('Time (s)', fontsize=11)
plt.ylabel('Voltage (V)', fontsize=11)
plt.grid(True, linestyle=':', alpha=0.6)
plt.tick_params(axis='both', labelsize=10)
plt.tight_layout()
plt.title('square-triangle')
plt.show()
```
