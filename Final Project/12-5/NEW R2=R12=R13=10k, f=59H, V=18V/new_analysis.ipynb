## Sine FFT and THD

```python
t = sin['TIME'].values
v = sin['CH1'].values

# === Signal and Sampling Info ===
N = len(t)
dt = np.mean(np.diff(t))
fs = 1 / dt
T = N * dt
print(f"Samples: {N}, Δt: {dt:.2e} s, fs: {fs:.1f} Hz, T: {T:.4f} s")

# === FFT Analysis ===
Vk = fft(v) / N
frequencies = np.fft.fftfreq(N, d=dt)
fk = frequencies[:N // 2]
Vk_mag = 2 * np.abs(Vk[:N // 2])  # one-sided magnitude spectrum

# === Manually detect fundamental: max FFT magnitude (excluding DC) ===
skip_bins = 1  # skip bin 0 (DC)
fundamental_idx = np.argmax(Vk_mag[skip_bins:]) + skip_bins
f1 = fk[fundamental_idx]
V1 = Vk_mag[fundamental_idx]
print(f"Manually detected fundamental: f1 = {f1:.2f} Hz (max FFT magnitude)")

# Find peak (fundamental)
peak_idx = np.argmax(Vk_mag)
f1_peak = fk[peak_idx]
mag_peak = Vk_mag[peak_idx]

# === THD Calculation (2nd–5th Harmonics) ===
harmonics = [f1 * n for n in range(2, 6)]
Vn_squared = 0
harmonic_labels = []

for n, h_freq in enumerate(harmonics, start=2):
    idx = np.argmin(np.abs(fk - h_freq))
    Vh = Vk_mag[idx]
    Vn_squared += Vh ** 2
    harmonic_labels.append((fk[idx], Vh, f"{n}f"))
    print(f"{n}f harmonic at {fk[idx]:.2f} Hz → |Vk| = {Vh:.4f}")

THD = np.sqrt(Vn_squared) / V1
print(f"\nTHD = {THD:.4f} ({THD*100:.2f}%)")

# === Optional: Inverse FFT ===
v_ifft = N * ifft(Vk)

# === PLOTS ===

# Time-domain signal
# plt.figure(figsize=(8, 3.5))
# plt.plot(t, v, color='#A2C2E0')
# plt.xlabel('Time (s)')
# plt.ylabel('Voltage (V)')
# plt.title('Input Signal')
# plt.grid(linestyle=':', alpha=0.4)
# plt.tight_layout()
# plt.show()

# Real and Imaginary FFT
plt.figure(figsize=(5, 3.5))
plt.scatter(fk, np.real(Vk[:N//2]), s=10, color='#9BCFF0', label='Re(Vk)')
plt.scatter(fk, np.imag(Vk[:N//2]), s=10, color='#FFB2E9', label='Im(Vk)')
plt.xlabel('Frequency (Hz)')
plt.ylabel('FFT Components')
plt.title('Real and Imaginary Parts of FFT')
plt.grid(linestyle=':', alpha=0.4)
plt.legend()
plt.xlim(0, 500)
plt.ylim(-15, 7)
plt.tight_layout()
plt.show()

# plots of magnitude spectrum
plt.figure(figsize=(5, 3.5))
# plt.scatter(fk, Vk_mag, s=10, color='#9BCFF0', label='')
plt.plot(fk, Vk_mag, color='#9BCFF0')

# Annotate the fundamental frequency
plt.annotate(f'Peak: {f1_peak:.2f} Hz',
             xy=(f1_peak, mag_peak),
             xytext=(f1_peak + 20, mag_peak + 2),
             arrowprops=dict(arrowstyle='->', color='gray'),
             fontsize=9)

plt.xlabel('Frequency (Hz)')
plt.ylabel('|Vk|')
plt.title('FFT Magnitude Spectrum')
plt.grid(linestyle=':', alpha=0.4)
# plt.legend()
plt.xlim(0, 1000)
plt.ylim(-3, 30)
plt.tight_layout()
plt.show()

# FFT Magnitude Spectrum with Harmonics
plt.figure(figsize=(9, 4))
plt.vlines(fk, 0, Vk_mag, color='#FFB2E9', lw=5)
plt.axvline(f1, color='#7D9FC1', linestyle='--', label=f'Fundamental: {f1:.1f} Hz')
for f_harm, mag, label in harmonic_labels:
    plt.annotate(label, xy=(f_harm, mag), xytext=(f_harm + 40, mag + 0.1),
                 arrowprops=dict(arrowstyle='->', color='gray'), fontsize=9)
plt.xlabel('Frequency (Hz)')
plt.ylabel('|Vk|')
plt.title('FFT Magnitude Spectrum with Harmonics')
plt.xlim(0, f1 * 6)
plt.ylim(0, 30)
plt.grid(linestyle=':', alpha=0.4)
plt.legend()
plt.tight_layout()
plt.show()

# Inverse FFT vs Original
plt.figure(figsize=(8, 3.5))
plt.plot(t, v, label='Original', color='#A2C2E0')
plt.plot(t, v_ifft.real, label='Inverse FFT', color='#FFB2E9', linestyle='--')
plt.xlabel('Time (s)')
plt.ylabel('Voltage (V)')
plt.title('Original vs Reconstructed Signal')
plt.legend()
plt.grid(linestyle=':', alpha=0.4)
plt.tight_layout()
plt.show()
```

This moderate distortion is consistent with the non-ideal diode shaping stage used to convert a triangle wave into an approximate sine. The presence of both even and odd harmonics reflects the nonlinear characteristics of the shaping circuitry.”

####
During extended operation, we observed that the ± voltage rails began to drift slightly apart—typically by 0.01–0.05 V. Since the rails were adjusted manually and independently, this introduced minor asymmetries in the clipping threshold of the diode shaper and affected the operating point of the op-amps. These small mismatches likely contributed to waveform distortion and the presence of even harmonics in the FFT spectrum, as confirmed by our THD analysis.

In FFT magnitude spectrum, we typically see:
- a fundamental frequency peak (the desired signal, 62.5Hz)
- Additional peaks at integer multiples of the fundamental: 2f, 3f, 4f,...
    - these are called harmonics and indicate distortion in your waveform

#### THD (Total Harmonic Distortion)
THD quantifies how much of your waveform's energy is in harmonics compared to the fundamental

$$
\text{THD}=\frac{\sqrt{V_2^2+V_3^2+V_4^2+\dots}}{V_1}
$$

-$V_1=$ magnitude of the fundamental
- $V_2,V_3,\dots=$ magnitudes of the 2nd, 3rd, etc. harmonics

Gives a single number that describes how pure your sine wave is"
- ideal sine wave: THD=0%
- realistic opamp sine: <5% is excellent, <10-15% is common
- want to explain imperfections due to opamp nonlinearity, filtering, clipping, or loading effects

How many harmonics?
- 3-5 harmonics is usually sufficient for a basic THD estimate
- beyond that, the FFT bins are in the noise floor unless your signal is very clean

## RMS sine fit

1. Define sine function model
$$
V(t)=A\sin(2\pi ft+\phi) + offset
$$

2. Estimate using initial parameters
- A (from peak values)
- f (from zero crossings or FFT)
- $\phi$ (0 to start)
- offset (mean of signal)

3. fit using `curve_fit`

4. calculate RMS value of fitted sine wave
$$
\text{RMS}=\frac{A}{\sqrt{2}}
$$

5. Plot original and fitted sine wave

```python
def sine(t, A, f, phi, offset):
    return A * np.sin(2 * np.pi * f * t + phi) + offset

# estimate initial guesses
A_guess = (np.max(v) - np.min(v)) / 2
offset_guess = np.mean(v)

# estimate frequency using FFT
sample_spacing = np.mean(np.diff(t))
fs = 1 / sample_spacing  # Sampling rate
n = len(v)
frequencies = np.fft.fftfreq(n, d=sample_spacing)
fft_vals = np.abs(np.fft.fft(v))
peak_freq = abs(frequencies[np.argmax(fft_vals[:n // 2])])  # Use only positive freqs
print(peak_freq)

# Initial guess: [Amplitude, Frequency, Phase, Offset]
initial_guess = [A_guess, peak_freq, 0, offset_guess]

# curve fitting
params, _ = curve_fit(sine, t, v, p0=initial_guess)
A_fit, f_fit, phi_fit, offset_fit = params

# find RMS voltage from amplitude
rms_voltage = A_fit / np.sqrt(2)

# fitted waveform
t_fit = np.linspace(t[0], t[-1], 1000)
v_fit = sine(t_fit, *params)

# plot measured + fitted dat
plt.figure(figsize=(8, 4))
plt.plot(t, v, label='Measured Signal', color='#A2C2E0', linewidth=2)
plt.plot(t_fit, v_fit, label='Sine Fit', color="#FDB9E5", linewidth=2)
plt.xlabel('Time (s)', fontsize=11)
plt.ylabel('Voltage (V)', fontsize=11)
plt.title('Sine Wave Fit')
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend(fontsize=10, loc='lower right')
plt.tight_layout()
plt.xlim(-0.03, 0.03)
plt.ylim(-50, 50)
plt.show()

# print resulting parameters
print(f"Amplitude (A): {A_fit:.4f} V")
print(f"Frequency (f): {f_fit:.2f} Hz")
print(f"Phase (phi): {phi_fit:.4f} rad")
print(f"Offset: {offset_fit:.4f} V")
print(f"RMS Voltage: {rms_voltage:.4f} V")
```

For report:

To quantify the signal's RMS voltage and assess its sinusoidal purity, we fitted a a sine wave to the measured output using nonlinear least squares. This allows us to directly estimate the amplitude, frequency, phase, and DC offset of the waveform even with mild distortion in the output.

We modeled the waveform data using the function $V(t)=A\sin(2\pi ft+\phi) +\text{offset}$. We estimated initial guesses for amplitude and offset from the peak-to-peak value and the mean, and the frequency from the dominant peak in the FFT spectrum. Then, we fit the model to our data using `scipy.optimize.curve_fit`.
